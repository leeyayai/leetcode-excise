```python
from collections import deque
from typing import Optional

class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right
class Solution:
    # 中序：左 -> 根 -> 右
    def inorderTraversal(self, root: TreeNode) -> list[int]:
        ans = []
        stk = []
        while stk or root:
            if root:
                stk.append(root)
                root = root.left
            else:
                root = stk.pop()
                ans.append(root.val)
                root = root.right
        return ans

    # 前序：根 -> 左 -> 右
    def preorderTraversal(self, root: TreeNode) -> list[int]:
        ans = []
        stk = []
        while stk or root:
            if root:
                ans.append(root.val)
                stk.append(root)
                root = root.left
            else:
                root = stk.pop()
                root = root.right
        return ans

    # 后序：左 -> 右 -> 根 【双栈经典写法】
    def postorderTraversal(self, root: TreeNode) -> list[int]:
        ans = []
        if not root:
            return ans
        stk1, stk2 = [root], []
        while stk1:
            root = stk1.pop()
            stk2.append(root.val)
            if root.left:
                stk1.append(root.left)
            if root.right:
                stk1.append(root.right)
        return stk2[::-1]

    # lst = [1, 2, 3]
    # lst[::-1]  # [3,2,1] 反转整个列表
    # lst[1::-1]  # [2,1] 从下标1，反向
    # lst[:-1]  # [1,2] 取到倒数第一个之前（去掉最后一个）
    # lst[-1]  # 3 拿最后一个元素

    #     递归形式
    def inorderTraversal1(self, root: TreeNode) -> list[int]:
        if not root:
            return []
        # 逻辑和上面Java完全等价（左→根→右），但是 ** 每次 + 都会生成新
        # list，存在拷贝开销 **。
        return self.inorderTraversal(root.left) + [root.val] + self.inorderTraversal(root.right)

        # 前序遍历：根 -> 左 -> 右

    def preorderTraversal2(self, root: Optional[TreeNode]) -> list[int]:
        res = []
        if not root:
            return res

        # dfs 是【内部嵌套函数】，不属于类方法，没有self
        def dfs(root: TreeNode):
            if root:
                res.append(root.val)
                dfs(root.left)
                dfs(root.right)

        # 直接调用 dfs，不用传self
        dfs(root)
        return res

    # 中序遍历：左 -> 根 -> 右
    def inorderTraversal2(self, root: Optional[TreeNode]) -> list[int]:
        res = []
        if not root:
            return res

        def dfs(root: TreeNode):
            if not root:
                return []
            dfs(root.left)
            res.append(root.val)
            dfs(root.right)

        dfs(root)
        return res

        # 后序遍历：左 -> 右 -> 根

    def postorderTraversal2(self, root: Optional[TreeNode]) -> list[int]:
        res = []
        if not root:
            return res

        def dfs(root: TreeNode):
            if not root:
                return []
            dfs(root.left)
            dfs(root.right)
            res.append(root.val)

        dfs(root)
        return res

    def maxDepth(self, root: Optional[TreeNode]) -> int:
        if not root:
            return 0
        L = self.maxDepth(root.left)
        R = self.maxDepth(root.right)
        return max(L, R) + 1

    def maxDepth_iter(self, root: Optional[TreeNode]) -> int:
        if not root:
            return 0
        # # 两种写法完全等价
        # q = deque([root])
        #
        # q = deque()
        # q.append(root)
        q = deque([root])
        depth = 0
        while q:
            depth += 1  # 进入新一层，深度+1
            for _ in range(len(q)):  # ,当前这一层一共有多少节点,一次性把当前层所有节点处理完
                root = q.popleft()  # 从队头取出节点,一次
                if root.left:
                    q.append(root.left)  # 左孩子入队
                if root.right:
                    q.append(root.right)  # 右孩子入队
        return depth + 1

    def invertTree(self, root: Optional[TreeNode]) -> Optional[TreeNode]:
        if not root:
            return None
        left = self.invertTree(root.left)
        right = self.invertTree(root.right)
        root.left, root.right = right, left
        return root

    def invertTree1(self, root: Optional[TreeNode]) -> Optional[TreeNode]:
        if not root:
            return None
        root.left, root.right = self.invertTree1(root.right), self.invertTree1(root.left)

    def invertTree_iter(self, root: Optional[TreeNode]) -> Optional[TreeNode]:
        if not root:
            return None
        q = deque([root])
        while q:
            root = q.popleft()
            root.left, root.right = root.right, root.left
            if root.left:
                q.append(root.left)
            if root.right:
                q.append(root.right)
        return root

    def isSymmetric(self, root: Optional[TreeNode]) -> bool:
        def dfs(left: TreeNode, right: TreeNode)->bool:
            if not left or not right:
                # // 两个只要有一个为空：返回 p == q（都空true，一空一非空false）
                return left == right
            return left.val == right.val and dfs(left.left, right.right) and dfs(left.right, right.left)
        return dfs(root.left, root.right)

    def diameterOfBinaryTree(self, root: Optional[TreeNode]) -> int:
        self.ans=0
        def dfs(root):
            # nonlocal  ans
            if not root:
                return 0
            L=dfs(root.left) # 获取该节点左子树的深度
            R=dfs(root.right) # 获取该节点右子树的深度
            self.ans=max(self.ans,L+R)
            # 不能去掉返回值，`return max(L,R)+1` 至关重要，不是多余的！
            return max(L,R)+1
        dfs(root)
        return self.ans

    # ans=0 #-- 写法错误
    def diameterOfBinaryTree1(self, root: Optional[TreeNode]) -> int:
        self.ans=0
        self.dfs(root)
        return self.ans

    def dfs(self,root:Optional[TreeNode])->int:
        if not root:
            return 0
        L=self.dfs(root.left)
        R=self.dfs(root.right)
        self.ans=max(L+R,self.ans)
        return max(L,R)+1

    def levelOrder(self, root: Optional[TreeNode]) -> list[list[int]]:
        if not root:
            return []
        q = deque([root])
        ans = []
        while q:
            cur_level = []
            for _ in range(len(q)):
                root = q.popleft()
                cur_level.append(root.val)
                if root.left:
                    q.append(root.left)
                if root.right:
                    q.append(root.right)
            ans.append(cur_level)
        return ans

    def sortedArrayToBST(self, nums: list[int]) -> Optional[TreeNode]:
        def dfs(left,right):
            if left>right:
                return None
            mid=(left+right)//2
            root=TreeNode(nums[mid])
            root.left=dfs(left,mid-1)
            root.right=dfs(mid+1,right)
            return root
        return dfs(0,len(nums)-1)

    def isValidBST(self,root:Optional[TreeNode])->bool:
        def dfs(node:Optional[TreeNode],min_node:Optional[TreeNode],max_node:Optional[TreeNode])->bool:
            if not node:
                return True
            if min_node and node.val<=min_node.val:
                return False
            if max_node and node.val>=max_node.val:
                return False
            return dfs(node.left,min_node,node) and dfs(node.right,node,max_node)
        return dfs(root,None,None)

    def kthSmallest(self,root:Optional[TreeNode],k:int)->int:
        res=[]
        def dfs(node:Optional[TreeNode]):
            if not node:
                return
            dfs(node.left)
            res.append(node.val)
            dfs(node.right)
        dfs(root)
        return res[k-1]

    def kthSmallest1(self, root: Optional[TreeNode], k: int) -> int:
        stk=deque()
        while True:
            while root:
                stk.append(root)
                root=root.left
            root=stk.pop()
            k-=1
            if(k==0):
                return root.val
            root=root.right
        return

    def kthSmallest2(self, root: Optional[TreeNode], k: int) -> int:
        stk=deque()
        while True:
            while root:
                stk.append(root)
                root=root.left
            root=stk.pop()
            k-=1
            if(k==0):
                break
            root=root.right
        return root.val

    def rightSideView(self, root: Optional[TreeNode]) -> list[int]:
        res=[]
        def dfs(node:Optional[TreeNode],depth:int):
            nonlocal res
            if not node:
                return
            if len(res)==depth:
                res.append(node.val)
            dfs(node.right,depth+1)
            dfs(node.left,depth+1)

        if not root:
            return []
        dfs(root,0)
        return res

    def flatten(self,root:Optional[TreeNode])->None:
        if not root:
            return
        self.flatten(root.left)
        self.flatten(root.right)
        old_right=root.right
        root.right=root.left
        root.left=None

        cur=root
        while cur.right:
            cur=cur.right
        cur.right=old_right


    def buildTree(self,preOrder:list[int],inorder:list[int])->Optional[TreeNode]:
        dict={v:i for i,v in enumerate(inorder)}
        def dfs(pre_order,in_order,pre_left,pre_right,in_left,in_right):
            if pre_left>pre_right:
                return None
            root=TreeNode(pre_order[pre_left])
            mid=dict[pre_order[pre_left]]
            size=mid-in_left
            root.left=dfs(pre_order,in_order,pre_left+1,pre_left+size,in_left,mid-1)
            root.right=dfs(pre_order,in_order,pre_left+size+1,pre_right,mid+1,in_right)
            return root
        return dfs(preOrder,inorder,0,len(preOrder)-1,0,len(inorder)-1)


    def lowestCommonAncestor(self,root:Optional[TreeNode],p:Optional[TreeNode],q:Optional[TreeNode])->Optional[TreeNode]:
        # 递归终止：空；或者当前节点就是p / q，直接返回自身
        if not root or root==p or root==q:
            return root
        left=self.lowestCommonAncestor(root.left,p,q)
        right=self.lowestCommonAncestor(root.right,p,q)
        if left is None:
            return right  # 左边没找到，那就在右边
        if right is None:
            return left  # 右边没找到，那就在左边
        return root  # 左右都没找到，说明p/q一个在左、一个在右，那root就是公共祖先

    def maxPathSum(self,root:Optional[TreeNode]):
        self.ans=float("-inf")
        def dfs(node:Optional[TreeNode])->int:
            if not node:
                return 0
            left = max(dfs(node.left),0) # 当前节点的左子树最大贡献
            right=max(dfs(node.right),0)  # 当前节点的右子树最大贡献
            curmax=node.val+left+right # 以当前节点为根的最大路径和
            self.ans=max(curmax,self.ans)
            return node.val+max(left,right) # 返回给父节点的贡献
        dfs(root)
        return self.ans










if __name__ == "__main__":
    # 先实例化Solution对象，再调用方法
    sol = Solution()

    root = TreeNode(-10, TreeNode(9), TreeNode(20, TreeNode(15), TreeNode(7)))
    print(sol.maxPathSum(root))  # expect 42

    '''
            1
          /   \
         2     7
        / \     \
       3   4     8
        \       /
         5     9
        /
       6
    '''
    n1 = TreeNode(1)
    n2 = TreeNode(2)
    n7 = TreeNode(7)
    n3 = TreeNode(3)
    n4 = TreeNode(4)
    n5 = TreeNode(5)
    n6 = TreeNode(6)
    n8 = TreeNode(8)
    n9 = TreeNode(9)

    n1.left, n1.right = n2, n7
    n2.left, n2.right = n3, n4
    n3.right = n5
    n5.left = n6
    n7.right = n8
    n8.left = n9

    # 找6和4的公共祖先，预期2
    anc = sol.lowestCommonAncestor(n1, n6, n4)
    print(anc.val)

    tree = TreeNode(1, TreeNode(2, TreeNode(3), TreeNode(4)), TreeNode(5, None, TreeNode(6)))
    sol.flatten(tree)
    # 遍历查看结果
    cur = tree
    res = []
    while cur:
        res.append(cur.val)
        cur = cur.right
    print(res)  # [1,2,3,4,5,6]

    #     1
    #    / \
    #   2   3
    #    \   \
    #     5   4
    tree = TreeNode(1,
                    TreeNode(2, None, TreeNode(5)),
                    TreeNode(3, None, TreeNode(4))
                    )
    print(sol.rightSideView(tree))  # [1,3,4]

    tree = TreeNode(3, TreeNode(1, None, TreeNode(2)), TreeNode(4))
    print(sol.kthSmallest1(tree, 2))  # 输出2，第2小

    n4 = TreeNode(4)
    n5 = TreeNode(5)
    n2 = TreeNode(2, n4, n5)
    n3 = TreeNode(3)
    root = TreeNode(1, n2, n3)
    sol = Solution()
    print(sol.diameterOfBinaryTree1(root))  # 3

    # build tree
    # 构造测试树：
    #        1
    #         \
    #          2
    #         /
    #        3
    n1 = TreeNode(1)
    n2 = TreeNode(2)
    n3 = TreeNode(3)
    n1.right = n2
    n2.left = n3

    sol = Solution()
    print("前序:", sol.preorderTraversal(tree))  # [1,2,3]
    print("中序:", sol.inorderTraversal(tree))  # [1,3,2]
    print("后序:", sol.postorderTraversal(tree))  # [3,2,1]
    print("最大深度:", sol.maxDepth(n1))

    # 对称树
    #        1
    #      /   \
    #     2     2
    #    / \   / \
    #   3  4  4  3
    n3_1 = TreeNode(3)
    n4_1 = TreeNode(4)
    left2 = TreeNode(2, n3_1, n4_1)

    n3_2 = TreeNode(3)
    n4_2 = TreeNode(4)
    right2 = TreeNode(2, n4_2, n3_2)

    root = TreeNode(1, left2, right2)
    sol = Solution()
    print(sol.isSymmetric(root))  # True
```
