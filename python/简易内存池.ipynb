```python
class MemSystem:
    def __init__(self, space: int):
        self.addr_map=dict()
        self.mem=[0]*space

    def request(self,size:int)->int:
        if size<=0:
            return -1
        consecutive=0
        mem_len=len(self.mem)
        for idx in range(mem_len):
            if  self.mem[idx]==0:
                consecutive+=1
            else:
                consecutive=0
            if consecutive>=size:
                start=idx-size+1
                self.mem[start:idx+1]=[1]*size
                self.addr_map[start]=size
                return start
        return -1

    def release(self,addr:int)->bool:
        if addr not in self.addr_map:
            return False
        size=self.addr_map[addr]
        self.mem[addr:addr+size]=[0]*size
        return True




if __name__=="__main__":
    ms=MemSystem(10)
    print(ms.request(3))
    print(ms.request(2))
    print(ms.release(0))
    print(ms.request(3))
```
