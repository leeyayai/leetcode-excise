```python
class Solution:
    def lower_bound(self,nums:list[int],target:int)->int:
        left,right=0,len(nums)-1 #闭区间[left,right]
        while left<=right:
            #循环不变量：第1个等于target的元素在[left,right]区间里
            #nums[left-1]<target
            #nums[right+1]>=target
            mid=(left+right)//2
            if nums[mid]<target:
                left=mid+1 # 范围缩小到[mid+1,right]
            else:
                right=mid-1 # 范围缩小到[left,mid-1]
        return left

    def lower_bound2(self,nums:list[int],target:int)->int:
        left=0
        right=len(nums) #开区间(left,right)
        while left<right:
            mid=(left+right)//2
            if nums[mid]<target:
                left=mid+1 # 范围缩小到(mid+1,right)
            else:
                right=mid # 范围缩小到(left,mid)
        return right

    def upper_bound(self,nums:list[int],target:int)->int:
        left = 0
        right = len(nums)  # 左闭右开区间[left,right)
        while left < right:
            mid = (left + right) // 2
            if nums[mid] < target:
                left = mid + 1  # 范围缩小到(mid+1,right)
            else:
                right = mid  # 范围缩小到(left,mid)
        return left

    # 搜索二维矩阵，展成一维，用二分查找
    def searchMatrix(self, matrix: list[list[int]], target: int) -> bool:
        m,n=len(matrix),len(matrix[0])
        left,right=0,m*n-1
        while left<=right:
            mid=(left+right)//2
            x=matrix[mid//n][mid%n]
            if x<target:
                left=mid+1
            else:
                right=mid-1
        # 先判断left不能超出总元素个数
        if left >= m * n:
            return False
        return matrix[left//n][left%n] == target

    def searchRange(self,nums:list[int],target:int)->list[int]:
        start=self.lower_bound(nums,target)
        if start==len(nums) or nums[start]!=target:
            return [-1,-1]
        # 如果start存在，那么end必定存在
        end=self.upper_bound(nums,target+1)-1
        return [start,end]

    # 搜索旋转排序数组
    # 不需要判断哪半边有序；**先找到旋转分割点，再二分其中一段**，思路清晰，避开 “判断哪半边有序” 的易错逻辑。
    # 旋转分割点：找最小值
    def findMin(self,nums:list[int])->int:
        # 返回最小值
        left, right = -1, len(nums) - 1  # 开区间 (-1, n-1)
        while left + 1 < right:  # 开区间不为空
            mid = (left + right) // 2
            if nums[mid] < nums[-1]:
                right = mid
            else:
                left = mid
        return right

    def search(self,nums:list[int],target:int)->int:
        n=len(nums)
        i=self.findMin(nums) # 找到旋转分割点
        if target>nums[-1]: # 前一个递增数组
            pos =self.lower(nums,0,i-1,target)
        else:
            pos =self.lower(nums,i,n-1,target)
        if pos<n and nums[pos]==target:
            return pos
        return -1

    def lower(self,nums:list[int],left:int,right:int,target:int)->int:
        while left<=right:
            mid=(left+right)//2
            if nums[mid]<target:
                left=mid+1
            else:
                right=mid-1
        return left











if __name__ == "__main__":
    sol=Solution()
    print(sol.lower_bound([1,2,3,4,5],2))
```
