```python
class Solution:
    def permute(self,nums:list[int])->list[list[int]]:
        n=len(nums)
        ans=[]
        temp=[]
        visited=[False]*n  #标记下标是否被选取

        def dfs():
            # 终止条件：临时列表长度等于数组长度，一组完整排列
            if len(temp)==n:
                # 等价：ans.append(temp[:])
                ans.append(temp.copy())
                return

            # 全排列：每一层循环全部下标，不从start开始
            for i in range(n):
                if visited[i]:
                    continue

                visited[i]=True
                temp.append(i)

                dfs()

                visited[i]=False
                temp.pop()

        return ans

    def subsets(selfself,nums:list[int])->list[list[int]]:
        n=len(nums)
        ans=[]
        temp=[]

        def dfs(start:int):
            # 进入当前层直接收集当前子集
            ans.append(temp[:])
            if len(temp)>=n:
                return

            # i从start开始，只向后取，不回头
            for i in range(start,n):
                temp.append(nums[i])
                dfs(i+1)
                temp.pop()
        dfs(0)
        return ans

    def letterCombinations(self,digits:str)->list[str]:
        if not digits:
            return []
        mapping = ["","","abc","def","ghi","jkl","mno","pqrs","tuv","wxyz"]
        ans=[]
        temp=[]

        def dfs(start:int):
            #  终止条件,处理完所有数字，收集结果
            if start==len(digits):
                ans.append("".join(temp))
                return
            # 当前数字对应的字符串
            chars=mapping[int(digits[start])]
            for c in chars:
                temp.append(c)
                dfs(start+1)
                temp.pop()

        dfs(0)
        return ans

    # candidates 元素可以**无限次重复选取**；
    # 数字全部为正数；求和等于 target 的所有组合；组合不能重复（`[2,3]`与`[3,2]`视为同一个）

# - 可以重复选（39 组合总和）：递归传 **i**，下一轮还能选自己
# - 不可以重复选（40 组合总和 Ⅱ /77 组合）：递归传 **i+1**，下一轮跳过当前下标
    def combinationSum(self,candidates:list[int],target:int)->list[list[int]]:
        ans=[]
        temp=[]

        def dfs(start:int,cur_sum:int):
            if cur_sum > target:
                return
            if cur_sum == target:
                ans.append(temp.copy())
                return
            for i in range(start,len(candidates)):
                dfs(i,cur_sum+candidates[i])
                temp.pop()

        dfs(0,0)
        return ans

    def generateParenthesis(self,n:int)->list[str]:
        ans=[]

        def dfs(s:str,l:int,r:int):
            # 左右括号都用完，收集答案
            if l==n and r==n:
                ans.append(s)
                return

            #剪枝：左超、右超、右>左、非法
            if l>n or r>n or r>l:
                return
            # 左+1，右+1
            dfs(s+"(",l+1,r)
            dfs(s+")",l,r+1)

        dfs("",0,0)
        return ans

    def exist(self,board:list[list[str]],word:str)->bool:
        m=len(board)
        n=len(board[0])
        words=list(word)

        def dfs(i:int,j:int,idx:int)->bool:
            # 越界、字符不匹配
            if i<0 or i>=m or j<0 or j>=n or board[i][j]!=words[idx]:
                return False
            #匹配到最后一位，成功
            if idx==len(words)-1:
                return True
            # 原地标记走过
            tmp=board[i][j]
            board[i][j]="."

            res = dfs(i+1,j,idx+1) or dfs(i-1,j,idx+1) or dfs(i,j+1,idx+1) or dfs(i,j-1,idx+1)

            # 回溯复原
            board[i][j]=tmp
            return res

        for i in range(m):
            for j in range(n):
                if dfs(i,j,0):
                    return True

        return False

    def partition(self,s:str)->list[list[str]]:
        n=len(s)
        ans=[]
        path=[]

        def is_pal(ss:str)->bool:
            l,r=0,len(ss)-1
            while l<r:
                if ss[l]!=ss[r]:
                    return False
                l+=1
                r-=1
            return True

        def dfs(start:int):
            if start>=n:
                ans.append(path.copy())
                return
            for i in range(start,n):
                sub=s[start:i+1]
                if is_pal(sub):
                    path.append(sub)
                    dfs(i+1)
                    path.pop()

        dfs(0)
        return ans
```
