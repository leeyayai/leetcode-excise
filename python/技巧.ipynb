任何数和0做异或运算，结果是自身，即 x ^ 0 = x。

任何数和其自身做异或运算，结果是0，即 x ^ x = 0。例如，5（101） ^ 5（101） = 0。

异或运算满足交换律和结合律，即 a ^ b ^ c = a ^ (b ^ c) = (a ^ b) ^ c。例如，5（101） ^ 3（011） ^ 4（100） = 5 ^ (3 ^ 4) = (5 ^ 3) ^ 4。

`sorted()` 是**返回新排序后的列表，不会原地改动 nums**。

对比：`nums.sort()` 才是原地排序。

```python
# 题目保证一定出现绝对众数 - 其出现次数大于其它所有数之和
    def majorityElement(self, nums: list[int]) -> int:
        arr=sorted(nums)
        return arr[len(nums)//2]
    
    def majorityElement1(self, nums: list[int]) -> int:
        nums.sort()
        return nums[len(nums)//2]
```

1. `nums[i]==0`：交换`nums[p0]`与`nums[i]`，`p0++，i++`。交换过来的是已经处理过的 1，可以 i 直接前进
2. `nums[i]==1`：不用交换，`i++`
3. `nums[i]==2`：交换`nums[p2]`与`nums[i]`，**只 p2--，i 不动！**

> 
> 交换过来的是还没检查的未知数字，i 不能 + 1，要再次判断

```python
# 颜色分类
    def sortColors(self,nums:list[int])->None:
        p0=i=0
        p2=len(nums)-1
        while i<=p2:
            if nums[i]==0:
                nums[p0],nums[i]=nums[i],nums[p0]
                p0+=1
                i+=1
            elif nums[i]==1:
                i+=1
            elif nums[i]==2:
                nums[i],nums[p2]=nums[p2],nums[i]
                p2-=1
```
