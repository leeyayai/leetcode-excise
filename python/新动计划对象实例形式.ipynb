```python
import re

class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

class Solution:
    def getNormalDeviceNum(self, start: int, end: int):
        count = end - start + 1
        for i in range(start, end + 1):
            s = str(i)
            if "4" in s or "18" in s:
                count -= 1
        return count

    def getNormalDeviceNum1(self, start: int, end: int):
        total=end-start+1
        bad=sum(1 for i in range(start,end+1) if "4" in str(i) or "18" in str(i))
        return total-bad

    def clusterLatency(self,arr: list[int])->int:
        arr.sort()
        mid=len(arr)//2
        ans=0
        for num in arr:
            ans+=abs(num-arr[mid])
        return ans

    def getDiffCnt(self,diff:int,arr:list[int])->int:
        s=set()
        for num in arr:
            s.add(num)
            s.add(num+diff)
        if diff==0:
            return 0
        return 2*len(arr)-len(s)

    def getDiffCnt1(self, diff: int, arr: list[int]) -> int:
        if diff == 0:
            return 0
        num_map={}
        count=0
        for num in arr:
            if (num+diff) in num_map:
                count+=1
        return count

    def getLongestVowelStrLen(self, input: str) -> int:
        slow,fast,maxLen=0,0,0
        n=len(input)
        vowels={'a','e','i','o','u','A','E','I','O','U'}
        while fast<n:
            c=input[fast]
            if c in vowels:
                maxLen=max(maxLen,fast-slow+1)
            else:
                slow+=1
            fast+=1
        return maxLen

    def getLongestFlawedVowelSubstrLen(self,flaw:int,s:str)->int:
        left,right=0,0
        maxLen,cnt=0,0
        n=len(s)

        def isYuan(c:str)->bool:
            vowels={'a','e','i','o','u','A','E','I','O','U'}
            return c in vowels

        while right<n:
            if not isYuan(s[right]):
                cnt+=1
            while cnt>flaw and left<right:
                if not isYuan(s[left]):
                    cnt-=1
                left==1

            if cnt==flaw and isYuan(s[right]) and isYuan(s[right]):
                maxLen=max(maxLen,right-left+1)
                right+=1
            return maxLen

    def getTheMinWithColMaxs(self, matrix: list[list[int]]) -> int:
        m=len(matrix)
        n=len(matrix[0])
        ans=[]
        for j in range(n):
            max_val=float('-inf')
            for i in range(m):
                max_val=max(max_val,matrix[i][j])
            ans.append(max_val)
        ans.sort()
        return ans[0]

    def getTheMinWithColMaxs1(self, matrix: list[list[int]]) -> int:
        # zip(*matrix) 按列打包
        col_max_list=[max(col) for col in zip(*matrix)]
        col_max_list.sort()
        return col_max_list[0]

    def transpose(self, inputStr: str) -> str:
        n=len(inputStr)
        i=1
        # 找到最小i满足 i*i >=n，必须i*i==n才合法
        while i*i<n:
            i+=1
        if i*i!=n:
            return "ERROR"
        res=[]
        for j in range(i):
            for k in range(j,1,i):
                res.append(inputStr[k*i+j])
        return "".join(res)

    def getMinStep(self,indentations:list[int])->int:
        ans,pre=0,0
        for cur in indentations:
            if cur>pre:
                ans+=cur-pre
            pre=cur
        return ans

    class Student:
        def __init__(self,number:int,height:int,weight:int):
            self.number=number
            self.height=height
            self.weight=weight

        def getNumber(self):
            return self.number

    def sortStudent(self,heights:list[int],weights:list[int])->list[int]:
        n=len(heights)
        students=[self.Student(i,heights[i],weights[i]) for i in range(n)]

    # 排序规则：
    # 优先身高升序；身高相同则体重升序；体重相同则编号升序
        students.sort(key=lambda x:(x.height,x.weight,x.number))

        ans=[student.getNumber() for student in students]
        return ans

    def sortStudent1(self, heights: list[int], weights: list[int]) -> list[int]:
        n=len(heights)
        students=[(i,heights[i],weights[i]) for i in range(n)]
        students.sort(key=lambda x:(x[1],x[2],x[0]))
        ans=[student[0] for student in students]
        return ans

    def regularID(self,cardIds:list[str])->list[str]:
        ans=[]
        for raw in cardIds:
            # 去掉空格，全部转小写
            s=raw.replace(" ","").lower()
            # 正则：^[a-zA-Z]([0-9]+) 首字母，后面跟着至少1个数字
            # 注意Java matches是全串匹配，re.fullmatch对应
            if re.fullmatch(r"^[a-zA-Z]([0-9]+)$",s):
                ans.append(s)
        return ans

    class Window:
        def __init__(self,l:int,r:int):
            self.l=l
            self.r=r

    def getBaseWindows(self,cap:int,visits:list[int])->Window:
        n=len(visits)
        preSum=[0]*(n+1)
        for i in range(n):
            preSum[i+1]=preSum[i]+visits[i]

        # 窗口长度从最长n往下遍历
        for length in range(n,0,-1):
            for l in range(n):
                r=l+length-1
                s=preSum[r+1]-preSum[l]
                if s<=cap:
                    return self.Window(l,r%n)
        return self.Window(-1,-1)

    def getMinSum(self,array1:list[int],array2:list[int],pairTarget:int)->int:
        m=len(array1)
        n=len(array2)
        ans=[]
        for i in range(m):
            for j in range(n):
                ans.append(array1[i]+array2[j])
        ans.sort()
        total=0
        for i in range(pairTarget):
            total += ans[i]
        return total

    def getMinSum1(self, array1: list[int], array2: list[int], pairTarget: int) -> int:
        all_sum=[a+b for a in array1 for b in array2]
        all_sum.sort()
        return sum(all_sum[:pairTarget])

    # def sortByDeleteOne(self, nums: list[int]) -> int:
    #     def is_ordered(a):
    #         inc=dec=True
    #         for p ,q in zip(a,a[1:]):




if __name__=="__main__":
    # 先实例化Solution对象，再调用方法
    sol = Solution()

    print(sol.getMinSum1([1, 3], [2, 4], 2))

    res = sol.getBaseWindows(10, [2, 3, 4, 1])
    print(res.l, res.r)

    print(sol.getNormalDeviceNum(1, 20))
    print(sol.clusterLatency([1,2,3,4,5]))
    print(sol.getDiffCnt(2, [1,2,3,4,5]))
    print(sol.getLongestVowelStrLen("aasiyyysjgA"))
    # mat= [[1,2,3],[4,5,6],[7,8,9]]
    mat = [
        [1, 2, 3],
        [4, 5, 6],
        [7, 8, 9]
    ]
    print(sol.getTheMinWithColMaxs(mat))
    print(sol.transpose("abcdefghi"))
    # 原字符串 a b c
    #           d e f
    #           g h i
    # 转置：a d g b e h c f i
    print(sol.getMinStep([1,2,3,4,5]))
    h = [170, 165, 170]
    w = [60, 50, 55]
    print(sol.sortStudent(h, w))
    # 预期输出：[2,3,1]
    print(sol.sortStudent1(h, w))
```
