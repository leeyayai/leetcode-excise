```python
from collections import deque, defaultdict


class Solution:
    def numIslands(self,grid:list[list[str]])->int:
        if not grid or not grid[0]:
            return 0
        m,n = len(grid),len(grid[0])
        res=0

        def dfs(i,j):
            if i<0 or i>=m or j<0 or j>=n or grid[i][j]=='0':
                return
            grid[i][j]='0'
            dfs(i+1,j)
            dfs(i-1,j)
            dfs(i,j-1)
            dfs(i,j+1)

        for i in range(m):
            for j in range(n):
                if grid[i][j]=='1':
                    res+=1
                    dfs(i,j)
        return res

    def orangesRotting(self,grid:list[list[int]])->int:
        M=len(grid)
        N=len(grid[0])
        q=deque()
        count=0

        for r in range(M):
            for c in range(N):
                if grid[r][c]==1:
                    count+=1
                if grid[r][c]==2:
                    q.append((r,c))

        round=0
        dirs=[(-1,0),(1,0),(0,-1),(0,1)]
        while count>0 and q:
            round+=1
            n=len(q)
            for _ in range(n):
                r,c=q.popleft()
                for dr,dc in dirs:
                    nr,nc=r+dr,c+dc
                    if 0<=nr<M and 0<=nc<N and grid[nr][nc]==1:
                        grid[nr][nc]=2
                        count-=1
                        q.append((nr,nc))
        return -1 if count>0 else round

    def canFinish(self,numCourses:int,prerequisites:list[list[int]])->bool:
        # in_degree[index]：课程index的入度
        in_degree=[0]*numCourses
        # adj邻接表：key为课程，value列表存所有后继课程（依赖key的课程）
        # adj={i:[] for i in range(numCourses)}
        adj=defaultdict(list)

        for a,b in prerequisites:
            # a依赖b，边 b -> a；a的入度+1
            in_degree[a]+=1
            adj[b].append(a)

        # 初始化队列：把所有入度为0的课程入队，可以直接
        q=deque()
        for i in range(numCourses):
            if in_degree[i]==0:
                q.append(i)
        while q:
            course = q.popleft()
            numCourses-=1 # 当前课程修完，剩余课程数量减一
            # 遍历所有依赖course的后继课程
            for nex_course in adj[course]:
                in_degree[nex_course]-=1 # 依赖course的课程的入度减一
                # 后继课程入度变成0，代表前置全部修完，可以入队列
                if in_degree[nex_course]==0:
                    q.append(nex_course)

        # numCourses等于0：全部课程可以修完，无环；否则存在环
        return numCourses == 0
```
