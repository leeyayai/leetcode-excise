```python
# 函数定义带`self` → 属于**类的实例方法**，必须依托类的实例调用，不能直接裸调用。
# 不带 class 包裹的全局函数，**不要写 self**。

def getNormalDeviceNum(start: int, end: int):
    count = end - start + 1
    for i in range(start, end + 1):
        s = str(i)
        if "4" in s or "18" in s:
            count -= 1
    return count

if __name__=="__main__":
    print(getNormalDeviceNum(1, 20))
```
