```python
from collections import deque


##   常数时间内检索到最小元素的栈
# - `stk`：普通栈，存原始入栈元素
# - `minStk`：辅助最小栈，**每一层记录当前栈状态下的最小值**

class Minstack:
    def __init__(self):
        self.stk=deque()
        self.min_stk=deque()

    def push(self, val)->None:
        self.stk.appendleft(val)
        if not self.min_stk:
            self.min_stk.appendleft(val)
        elif val<=self.min_stk[0]:
            self.min_stk.appendleft(val)
        else:
            self.min_stk.appendleft(self.min_stk[0])

    def pop(self)->None:
        self.stk.popleft()
        self.min_stk.popleft()

    def top(self)->int:
        return self.stk[0]

    def getMin(self)->int:
        return self.min_stk[0]
```
