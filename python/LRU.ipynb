```python
class Node:
    def __init__(self,key:int,value:int):
        self.key=key
        self.value=value
        self.pre=None
        self.next=None

class LRUCache:
    def __init__(self,capacity:int):
        self.capacity=capacity
        self.map={}
        self.head=Node(0,0)
        self.tail=Node(0,0)
        self.head.next=self.tail
        self.tail.pre=self.head

    def _insert_head(self,node:Node):
        nex=self.head.next
        self.head.next=node
        node.pre=self.head
        node.next=nex
        nex.pre=node

    def _cut_node(self,node:Node):
        pre_node=node.pre
        next_node=node.next
        pre_node.next=next_node
        next_node.pre=pre_node
        node.pre=None
        node.next=None

    def _del_node(self)->Node:
        cur=self.tail.pre
        last=cur.pre
        last.next=self.tail
        cur.pre=None
        cur.next=None
        return cur

    def get(self,key:int)->int:
        if key in self.map:
            node=self.map[key]
            self._cut_node(node)
            self._insert_head(node)
            return node.value
        return -1

    def put(self,key:int,value:int):
        if key in self.map:
            node=self.map[key]
            node.value=value
            self._cut_node(node)
            self._insert_head(node)
        else:
            node =Node(key,value)
            self._insert_head(node)
            self.map[key]=node
            if self.capacity<len(self.map):
                self._del_node()
                self.map.pop(node.key)


# =========测试样例========
if __name__ == "__main__":
    cache = LRUCache(2)
    cache.put(1, 1)
    cache.put(2, 2)
    print(cache.get(1))   # 1
    cache.put(3, 3)
    print(cache.get(2))   # -1
    cache.put(4, 4)
    print(cache.get(1))   # -1
    print(cache.get(3))   # 3
    print(cache.get(4))   # 4
```
