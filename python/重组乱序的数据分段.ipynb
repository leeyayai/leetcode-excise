```python
class DataFragment:
    def __init__(self, dataType:int,index:int,endFlag:int,len:int,data:str):
        self.dataType = dataType
        self.index = index
        self.endFlag = endFlag
        self.len = len
        self.data = data
    def __repr__(self):
        return f'[{self.dataType},{self.index},{self.endFlag},{self.len},"{self.data}"]'

class Solution:
    def recombine(self, dataFragments: list[DataFragment], dstType: int) -> str:
        list_frag=[]
        idx_sex=set()
        for frag in dataFragments:
            if frag.dataType==dstType:
                if frag.index not in idx_sex:
                    list_frag.append(frag)
                    idx_sex.add(frag.index)

        if len(list_frag)==0:
            return "NA"

        # 按index升序排序
        list_frag.sort(key=lambda x: x.index)
        # 判断：最大index +1 是否等于片段数量，索引必须连续从0开始无空缺
        if len(list_frag) != list_frag[-1].index + 1:
            return "NA"

        # 检查是否存在 endFlag ==1
        has_end=False
        for frag in list_frag:
            if frag.endFlag==1:
                has_end=True

        if has_end:
            # 再次排序（Java源码重复排序，原样保留）
            list_frag.sort(key=lambda x: x.index)
            sb = []
            for frag in list_frag:
                sb.append(frag.data)
            return "".join(sb)
        else:
            return "NA"

    def recombine1(self, dataFragments: list[DataFragment], dstType: int) -> str:
        frags_list=[]
        seen_idx=set()
        for f in dataFragments:
            if f.dataType==dstType and f.index not in seen_idx:
                frags_list.append(f)
                seen_idx.add(f.index)
        if not frags_list:
            return "NA"

        frags_list.sort(key=lambda x:x.index)
        if len(frags_list)!=frags_list[-1].index+1:
            return "NA"
        if any(f.endFlag==1 for f in frags_list):
            frags_list.sort(key=lambda x:x.index)
            return "".join(f.data for f in frags_list)
        return "NA"

if __name__ == "__main__":
    frags = [
        DataFragment(1, 0, 0, 2, "ab"),
        DataFragment(1, 1, 1, 2, "cd"),
    ]
    sol = Solution()
    print(sol.recombine1(frags, 1))  # abcd
```
