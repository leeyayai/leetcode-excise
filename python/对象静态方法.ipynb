```python
class Solution:
    @staticmethod
    def getNormalDeviceNum(start: int, end: int):
        count = end - start + 1
        for i in range(start, end + 1):
            s = str(i)
            if "4" in s or "18" in s:
                count -= 1
        return count

if __name__=="__main__":
    print(Solution.getNormalDeviceNum(1, 20))
```
