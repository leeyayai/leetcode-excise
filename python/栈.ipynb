```python
from collections import deque


class Solution:
    def isValid(self,s:str)->bool:
        stack=[]
        for char in s:
            if char == '(':
                stack.append(')')
            if char == '{':
                stack.append('}')
            if char == '[':
                stack.append(']')
            else:
                if not stack or char != stack.pop():
                    return False
            return len(stack)==0

    #字符串解码

    def decodeString(self,s:str)->str:
        ans=[]
        stk_str=deque()
        stk_time=deque()
        time=0
        for c in s:
            if c.isdigit():
                time=time*10+int(c)
            elif c=='[':
                stk_str.appendleft(''.join(ans))
                ans=[]
                stk_time.appendleft(time)
                time=0
            elif c==']':
                #"".join(ans) 字符列表 → 字符串
                #`list(xxx)`：字符串 → 字符列表
                ans=list(stk_str.popleft()+"".join(ans)*stk_time.popleft())
            else:
                ans.append(c)
        return ''.join(ans)

    # 每日温度 -- 单调栈
    # 栈顶就是 ** 比当前i温度更高的最近的右边下标 **
    def dailyTemperatures(self,temperatures:list[int]):
        n=len(temperatures)
        ans=[0]*n
        stk=deque()
        for i in range(n-1,-1,-1):
            while stk and temperatures[stk[0]]<=temperatures[i]:
                # 栈顶温度 小于 当前温度
                stk.popleft()
            if stk:
                ans[i]=stk[0]-i
            stk.appendleft(i)
        return ans











if __name__=="__main__":
    sol=Solution()
    print(sol.isValid("()"))  # True
    print(sol.isValid("()[]{}"))  # True
    print(sol.isValid("(]"))  # False
    print(sol.isValid("["))  # False

    print(sol.decodeString("3[a2[c]]"))  # abcbcabcbcabcbc
    print(sol.decodeString("2[abc]3[cd]ef"))  # abcabc cdcdcd ef
```
