```python
class Solution:
    def climbStairs(self,n:int)->int:
        if n<=1:
            return 1
        dp=[0]*(n+1)
        dp[0]=1
        dp[1]=1
        # `dp[i]`：到达第`i`阶的方法数
        # 拓展：如果一次可以走 1,2,3 阶，dp [i]=dp [i‑1]+dp [i‑2]+dp [i‑3]。
        for i in range(2,n+1):
            dp[i]=dp[i-1]+dp[i-2]
        return dp[n]

    # 在「杨辉三角」中，每个数是它左上方和右上方的数的和。
    # 每行首尾为 1；中间值等于上一行相邻两数之和；
    # 行号 i，该行一共有`i+1`个元素，j 循环范围`0~i`；
    # `ans.get(i‑1)`拿到上一行。
    def generate(self,numRows:int)->list[list[int]]:
        ans=[[1]]
        for i in range(1,numRows):
            temp=[]
            for j in range(i+1):
                if j==0 or j==i:
                    temp.append(1)
                else:
                    temp.append(ans[i-1][j-1]+ans[i-1][j])
            ans.append(temp)
        return ans

    # 打家劫舍，你是小偷，房屋排成一排，**不能偷相邻两家**，求能偷到的最大金额。
    # `nums[i]`是第 i 家钱数。
    # `dp[i]`：偷到第 i 家时能获得的最大金额。 前`i`间房子，可以偷到的最大总金额。
    # `dp[i]=max(dp[i-1],dp[i-2]+nums[i])`：偷或不偷第`i`家。 -- 状态转移：当前位置两种选择：偷 / 不偷；不能相邻；
    # `dp[i-1]`：不偷第`i`家，前`i-1`家的最大金额。
    # `dp[i-2]+nums[i]`：偷第`i`家，前`i-2`家的最大金额 + 第`i`家的金额。
    def rob(self,nums:list[int])->int:
        n=len(nums)
        if n==1:
            return nums[0]
        dp=[0]*n
        dp[0]=nums[0]
        dp[1]=max(nums[0],nums[1])
        for i in range(2,n):
            dp[i]=max(dp[i-1],dp[i-2]+nums[i])
        return dp[-1]

    # 完全平方数 给定正整数 `n`，找出**最少**数量的完全平方数（1,4,9,16…）之和等于 n，返回这个最少个数。 。
    # 输入：n = 12
    # 输出：3
    # 解释：12 = 4 + 4 + 4
    # 完全背包：DP 定义：dp [i] 凑 i 的最少平方数数量，初始化：dp [i]=i，最坏全部取 1；dp [0]=0；
    # 转移方程：\(dp[i]=\min(dp[i],\ dp[i-j^2]+1)\)
    # `j*j` 就是当前选的平方数，选了它，就等价于：凑 `i-j²` 的方案 + 1 个当前平方数。
    # **完全背包**，物品是平方数，可以无限选；背包容量是 n；求装满背包的**最少物品数量**。
    def numSquares(self,n:int)->int:
        dp=[i for i in range(n+1)]
        for i in range(1,n+1):
            j=1
            while j*j<=i:
                dp[i]=min(dp[i],dp[i-j*j]+1)
                j+=1
        return dp[n]

    # 01 背包：每件物品只能选 1 次 → 背包维度倒序遍历
    # 完全背包：物品可以无限重复选 → 背包维度正序遍历
    # 输入：coins = [1, 2, 5], amount = 11
    # 输出：3
    # 解释：11 = 5 + 5 + 1
    # 零钱兑换：物品是现成数组 coins，外层 for 遍历硬币；
    # 完全平方数：没有预先把平方数存数组；外层循环是背包 i，内层循环动态生成平方物品`j*j`。

    # dp[j]：凑成总金额 j 所需的最少硬币个数。初始化为一个较大的数，表示初始状态不可达。dp[0]=0，凑成金额 0 需要 0 个硬币。
    def coinChange(self,coins:list[int],amount:int)->int:
        INF=float('inf')
        dp=[INF]*(amount+1)
        dp[0]=0
        for c in coins:
            # 完全背包，物品可以无限选，外层遍历物品，内层正序遍历背包容量
            for j in range(c,amount+1):
                dp[j]=min(dp[j],dp[j-c]+1)
        return dp[amount] if dp[amount]!=INF else -1

    # 上面两个都是完全背包，分割等和子集（经典 01 背包）
    # 给一个数组，能不能选一部分元素，使得选出数字之和 = 总和的一半。每个数字**只能选一次**（01 背包核心）
    # dp[j]：背包容量j，是否可以凑出和为j。
    # 物品：数组里每个数字 nums [i]，重量 = 价值 = nums [i]，只能选一次
    # 背包：容量为 sum//2
    def canPartition(self,nums:list[int])->bool:
        total=sum(nums)
        if total%2 != 0:
            return False
        target=total//2
        dp=[False]*(target+1)
        dp[0]=True

        # 背包DP的本质是对每个物品独立做选或不选的决策，物品之间没有依赖，遍历物品的先后顺序不会改变最终dp结果，
        # 所以外层物品数组不需要排序。排序只在贪心或者剪枝优化的时候才用到。
        for num in nums:
            # 01背包:倒序，range(target,num-1,-1)
            for j in range(target,num-1,-1):
                dp[j]=dp[j] or dp[j-num]
        return dp[target]

    # 最后一块石头的重量：一堆石头，每次拿两块相撞，剩下重量差。求最后**最小可能剩下的石头重量**
    # 核心转化：把石头分成两组，两组重量尽量接近。目标：找到不超过`sum/2`的最大重量。
    # `dp[j]`：背包容量 j，最多能装多少石头重量。
    def lastStoneWeight(self,stones:list[int])->int:
        total=sum(stones)
        target=total//2
        dp=[0]*(target+1)
        for s in stones:
            # 01背包
            for j in range(target,s-1,-1):
                dp[j]=max(dp[j],dp[j-s]+s)
        return total-2*dp[target]







    def test(self):
        coins=[]
        amount=0
        dp=[]
        # 零钱兑换 完全背包（物品无限拿，j 正序）
        for c in coins:
            # `range(c, amount+1)` → j: c, c+1, c+2 ... amount  **正序，j++**
            for j in range(c, amount + 1):
                dp[j] = min(dp[j], dp[j - c] + 1)

        nums=[]
        target=0
        # 分割等和子集 01 背包（物品只能拿 1 次，j 倒序）
        for num in nums:
            # j从大到小：range(target, num-1, -1)
            for j in range(target, num - 1, -1):
                # `range(target, num-1, -1)` → j: target, target-1 ... num **倒序，j--**
                dp[j] = dp[j] or dp[j - num]

    #
```
