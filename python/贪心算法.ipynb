```python
class Solution:
    def maxProfix(self,prices:list[int]) -> int:
        if len(prices)<1:
            return 0
        left=0
        ans=0
        # `left`：记录到当前为止**最低的股价（买入点）**
        # `right`：遍历每一天作为卖出时机
        for right in range(1,len(prices)):
            temp = prices[right]-prices[left]
            if temp<0:
                left=right
            ans=max(ans,temp)
        return ans

    # `max_reach`：记录当前能够到达的 ** 最远下标 **。
    # `nums[i]+i`：从位置`i`可以跳到的最远位置。
    def canJump(self,nums:list[int])->bool:
        n=len(nums)
        max_reach=0
        i=0
        while i<= max_reach:
            max_reach=max(max_reach,nums[i]+i)
            if max_reach>=n-1:
                return True
            i+=1
        return False

    def jump(self,nums:list[int])->int:
        n=len(nums)
        if n<=1:
            return 0
        count=0
        cur_max=0
        reach=0
        for i in range(n-1):
            reach=max(reach,i+nums[i])
            if i == cur_max:
                cur_max=reach
                count+=1
        return count

    # 对给定字符串划分为尽可能多的片段，同一字母最多出现在一个片段
    def partition(self,s:str)->list[int]:
        last=[0]*26
        for idx ,c in enumerate(s):
            # `idx`：这个字符**最后出现的位置**，赋值给数组对应位置
            last[ord(c)-ord('a')]=idx
        res=[]
        far=0
        start=-1
        for i,c in enumerate(s):
            far=max(far,last[ord(c)-ord('a')])
            if i == far:
                res.append(far-start)
                start=i
        return res
```
