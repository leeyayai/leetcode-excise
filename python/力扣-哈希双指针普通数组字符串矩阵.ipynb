```python
class Solution:
    def minWindow(self, s: str, t: str) -> str:
        dic = defaultdict(int)
        l=0
        cur = len(t)  # 当前还需要补齐的字符个数
        
        for ch in t:
            sic[ch] += 1
            
        ans = ""  # 存最短子串
        
        for r,c in enumerate(s)
            dic[c] -= 1
            # 只有原本就是需要的字符，消耗掉才减少缺口
            if dic[c] >=0:
                cur -= 1
                
            # cur == 0: 当前窗口[l,r]已经完全覆盖t
            while cur ==0:
                # 更新最短
                if not ans or (r-1+1)<len(ans):
                    ans=s[l:r+1]
                    
                # 移出左边界字符
                left_char = s[1]
                dic[left_char+=1]
                # 一旦这个字符缺口>0，窗口不再合法
                if dic[left_char] > 0
                    cur += 1
                l += 1
        return ans
        
        
```

```python
def merge(self,intervals:List[List[int]])->List[List[int]]:
    if not intervals:
        return []
    # 按区间左端点排序，等价 java: Arrays.sort(intervals,(a,b)->a[0]-b[0])
    intervals.sort(key=lambda x :x[0])
    res=[intervals[0]]
    m=len(intervals)
    index=1
    for i in range(1,m):
        if intervals[i][0] <= intervals[i-1][1]:
            right = max(intervals[i-1][1],intervals[i][1])
            res.pop(index-1)
            res.append([intervals[i-1][0]],right)
            intervals[i][0]=intervals[i-1][0]
            intervals[i][1]=right
        else:
            res.append(intervals[i])
            index += 1
            
    return res
```

```python
class Solution:
    def firstMissingPositive(self, nums: list[int]) -> int:
        n = len(nums)
        for i in range(n):
            # 如果当前学生的学号在 [1,n] 中，但（真身）没有坐在正确的座位上
            while 1 <= nums[i] <= n and nums[nums[i] - 1] != nums[i]:
                # 那么就交换 nums[i] 和 nums[j]，其中 j 是 i 的学号
                j = nums[i] - 1  # 减一是因为数组下标从 0 开始
                nums[i], nums[j] = nums[j], nums[i]

        # 找第一个学号与座位编号不匹配的学生
        for i in range(n):
            if nums[i] != i + 1:
                return i + 1

        # 所有学生都坐在正确的座位上
        return n + 1
```

```python
# 算法原理（原地 O (1) 额外空间）
def setZeroes(self, matrix: List[List[int]]) -> None:
    
    m = len(matrix)
    n = len(matrix[0])
        
    row_flag = False   # 标记第 0 列是否原本存在0
    col_flag = False   # 标记第 0 行是否原本存在0
    
    # 检查第一列
    for i in range(m):
        if matrix[i][0]==0
        row_flag=True
        break
        
    for i in range(n):
        if matrix[0][j]==0
            col_flag=True
            break
    
    # 使用第0行、第0列作为标记数组
    for i in range(1,m):
        for j in range(1,n):
            if matrix[i][j]==0:
                matrix[i][0]=0
                matrix[0][j]=0
                
    # 根据标记，内部元素置零
    for i in range(1,m):
        for j in range(1,n):
            if matrix[i][0]==0 or matrix[0][j]==0:
                matrix[i][j]==0
                
        # 处理第一行
    if col_flag:
        for j in range(n):
            matrix[0][j] = 0
        
    # 处理第一列
    if row_flag:
        for i in range(m):
            matrix[i][0] = 0    
    
    
    
```

```python
def spiralOrder(self, matrix: List[List[int]]) -> List[int]:
    m=len(matrix)
    n=len(matrix[0])
    upon=0
    low=m-1
    left=0
    right=n-1
    ans=[]
    
    while True:
        # 左 → 右，遍历上边
        for j in range(left,right+1):
            ans.append(matrix[upon][j])
        upon+=1
        if upon >low
            break
            
        # 上 → 下，遍历右边
        for i in range(upon,low+1):
            ans.append(matrix[i][right])
        right-=1
        if right  < left:
            break
            
        # 右 → 左，遍历下边
        for j in range(right,left-1,-1):
            ans.append(matrix[low][j])
        low-=1
        if low<upon
            break
            
        # 下 → 上，遍历左边
        for i in range(low,upon-1,-1):
            ans.append(matrix[i][left])
        left+=1
        if left>right
            break
            
    return ans
        
        
        
    
    
```
