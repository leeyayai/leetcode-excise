```python
class Operation:
    def __init__(self,opType:str,number:str):
        self.opType = opType
        self.number = number

    def __repr__(self):
        return f"[{self.opType},\"{self.number}\"]"

class MyRecord:
    def __init__(self,number:str,accepted:int,refused:int):
        self.number = number
        self.accepted = accepted
        self.refused = refused

    def __repr__(self):
        return f"[\"{self.number}\",{self.accepted},{self.refused}]"

class Solution:
    def getPhoneRecords(self,operations: list[Operation])->list[MyRecord]:
        white_set=set()
        prefix_list=[]
        call_map={}

        for op in operations:
            if op.opType=='W':
                s=op.number
                if s[-1]=="*":
                    prefix_list.append(s[:-1])
                else:
                    white_set.add(s)
            elif op.opType=='C':
                num=op.number
                if num not in call_map:
                    call_map[num] = MyRecord(num,0,0)
                rec=call_map[num]
                ok=num in white_set
                if not ok:
                    for p in prefix_list:
                        if num.startswith(p):
                            ok=True
                            break
                if ok:
                    rec.accepted+=1
                else:
                    rec.refused+=1
        return list(call_map.values())

if __name__=="__main__":
    ops = [
        Operation('W', "138*"),
        Operation('W', "13900001111"),
        Operation('C', "13812345678"),
        Operation('C', "13900001111"),
        Operation('C', "13788889999"),
    ]
    sol = Solution()
    res = sol.getPhoneRecords(ops)
    for r in res:
        print(r)
```
